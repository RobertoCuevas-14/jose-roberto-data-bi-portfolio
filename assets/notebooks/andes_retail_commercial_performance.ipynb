KPIs incluidos

Ingresos: permite conocer el valor total generado por las ventas.

Margen: permite evaluar qué porcentaje de los ingresos se conserva como ganancia.

### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):**  
Andes Retail Group registró ingresos totales por $5,531,994 durante el periodo 2024–2025, generando una ganancia de $1,941,749 y un margen global de 35.1 %. En total se procesaron 5,000 pedidos, con un ticket promedio de $1,106.

**C (Complicación):**  
El desempeño no fue uniforme durante todo el periodo. Los ingresos presentaron variaciones mensuales y la contribución económica se concentró principalmente en determinados países y categorías. Esto hace necesario identificar cuáles componentes explican el resultado y dónde existen oportunidades de mejora.

**Q (Pregunta):**   
¿Cómo evolucionó el desempeño comercial y qué países y categorías explican principalmente los ingresos y la ganancia del negocio?

**A (Respuesta):**  
Perú fue el país con la mayor ganancia acumulada, mientras que Deportes concentró el mayor nivel de ingresos. La evolución mensual mostró picos al inicio de los años y niveles más bajos en algunos meses intermedios. En general, el negocio mantuvo un margen sólido de 35.1 %, aunque las diferencias en resultados estuvieron principalmente asociadas al volumen de ventas.

#### 🔎 Vista Detalle

**S (Situación):**  
El análisis detallado muestra que los márgenes son similares entre Perú, Chile y Colombia, ubicándose aproximadamente entre 34.9 % y 35.8 %. También se observan diferencias importantes entre segmentos de clientes y estaciones del año.

**C (Complicación):**  
Aunque la rentabilidad porcentual se mantiene estable entre los países, los ingresos y la ganancia se concentran en el segmento Premium y durante la temporada de Verano. El segmento Económico y la temporada de Invierno presentan una contribución considerablemente menor.

**Q (Pregunta):**  
¿Qué segmentos y periodos generan mayor valor para el negocio y dónde existen oportunidades para mejorar el desempeño?

**A (Respuesta):**  
El segmento Premium combina el mayor volumen de ingresos con el ticket promedio más alto, por lo que representa el principal generador de valor. Verano presenta el mejor desempeño estacional, mientras que Invierno registra los niveles más bajos. Se recomienda proteger la propuesta de valor del segmento Premium, reforzar las estrategias comerciales durante Verano y evaluar acciones específicas para incrementar la participación del segmento Económico y mejorar el desempeño de Invierno.
